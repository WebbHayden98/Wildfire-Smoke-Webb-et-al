```python
def cleanaod(data):
    
    time = np.array(data['Date(dd:mm:yyyy)'])
    aod = np.array(data['AOD_500nm'])
    aod[np.where(aod==-999)] = np.nan
    
    data_month = np.ones_like(time)*np.nan
    data_year = np.ones_like(time)*np.nan
    data_day = np.ones_like(time)*np.nan
    
    for i in range(len(time)):    
        data_day[i]=(int(time[i][0:2]))
        data_month[i]=(int(time[i][3:5]))
        data_year[i]=(int(time[i][6:10]))
        
    aod = aod[np.where((data_year<=2022)&(data_year>=2011))[0]]
    time = time[np.where((data_year<=2022)&(data_year>=2011))[0]]
    data_day = data_day[np.where((data_year<=2022)&(data_year>=2011))[0]]
    data_month = data_month[np.where((data_year<=2022)&(data_year>=2011))[0]]
    data_year = data_year[np.where((data_year<=2022)&(data_year>=2011))[0]]
    return time, aod, data_day, data_month, data_year
```
