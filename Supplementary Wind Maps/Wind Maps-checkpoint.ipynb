```python
x = Day.shape[0]
newx = int(x/24)

Days = np.ones(newx)*np.nan
Months = np.ones(newx)*np.nan
Years = np.ones(newx)*np.nan
Dailyu500 = np.ones((newx,241,361))*np.nan
Dailyv500 = np.ones((newx,241,361))*np.nan
#Dailyu700 = np.ones((newx,201,481))*np.nan
#Dailyv700 = np.ones((newx,201,481))*np.nan

for i in range(0,x,24):
    q=int(i/24)
    Dailyu500[q] = (np.nanmean(u500[i+10:i+14],axis=0))
    Dailyv500[q] = (np.nanmean(v500[i+10:i+14],axis=0))
    #Dailyu700[q] = ((u700[i] + u700[i+1] + u700[i+2] + u700[i+3])/4)
    #Dailyv700[q] = ((v700[i] + v700[i+1] + v700[i+2] + v700[i+3])/4)
    Days[q] = Day[i]
    Months[q] = Month[i]
    Years[q] = Year[i]
```

```python
geowind500 = np.ones_like(Dailyu500)*np.nan
#geowind700 = np.ones_like(Dailyu700)*np.nan
for i in range(newx):
    geowind500[i] = np.sqrt(Dailyu500[i]**2 + Dailyv500[i]**2)
    #geowind700[i] = np.sqrt(Dailyu700[i]**2 + Dailyv700[i]**2)
```

```python
x = Day.shape[0]
newx = int(x/24)

Days = np.ones(newx)*np.nan
Months = np.ones(newx)*np.nan
Years = np.ones(newx)*np.nan
DailyhPa500 = np.ones((newx,241,361))*np.nan
DailyhPa700 = np.ones((newx,241,361))*np.nan

for i in range(0,x,24):
    q=int(i/24)
    DailyhPa500[q] = (np.nanmean(hPa500[i+10:i+14],axis=0))
    #DailyhPa700[q] = ((hPa700[i] + hPa700[i+1] + hPa700[i+2] + hPa700[i+3])/4)
    Days[q] = Day[i]
    Months[q] = Month[i]
    Years[q] = Year[i]
```

```python
lon2,lat2 = np.meshgrid(lon[:,0],lat[0,:])

fig, ax = plt.subplots(1, 1, subplot_kw={'projection': ccrs.PlateCarree()},figsize=(12,8))
data_crs = ccrs.PlateCarree()

extent = [-140,-50,20,70]

M = 8
D = 2
Y = 2021

height = DailyhPa500[np.where((Months==M)&(Days==D)&(Years==Y))][0]
wind = geowind500[np.where((Months==M)&(Days==D)&(Years==Y))][0]
u = Dailyu500[np.where((Months==M)&(Days==D)&(Years==Y))][0]
v = Dailyv500[np.where((Months==M)&(Days==D)&(Years==Y))][0]



ax._autoscaleXon = False
ax._autoscaleYon = False
contours = ax.contour(lon2,lat2,height/100,15,colors='black',linewidths=1,transform=data_crs)
#contours = ax.contourf(lon2,lat2,height,200,linewidths=1,transform=data_crs)
plt.clabel(contours, inline = True, fontsize=8, fmt='%d', colors = 'black')

cf = ax.contourf(lon2, lat2, wind,levels=np.arange((0),(32),2), cmap='YlOrRd', transform=data_crs)
cb = plt.colorbar(cf, orientation='horizontal', pad=0, aspect=50)
ax.add_feature(cfeature.COASTLINE,linewidth=0.5)
ax.add_feature(cfeature.BORDERS,linewidth=0.5)
ax.add_feature(cfeature.STATES,linewidth=0.5)
ax.set_extent(extent,crs=data_crs)
fig.tight_layout(pad=0.5)

wind_slice = (slice(None, None, 25), slice(None, None, 25))
ax.barbs(lon2[wind_slice[0],wind_slice[1]], lat2[wind_slice[1],wind_slice[0]],
         u[wind_slice[0],wind_slice[1]],
         v[wind_slice[0],wind_slice[1]],
         pivot='middle', color='black', transform=data_crs)

plt.title('500hPa Geopoential Winds (' + str(M) + '/' + str(D) + '/' + str(Y) + ')',name='Calibri',weight='bold',\
             style='italic')
cb.ax.set_xlabel('Velocity [m/s]',weight='bold',name='Calibri')

plt.savefig(str(M)+str(D)+str(Y)+'.png', bbox_inches='tight')
```
